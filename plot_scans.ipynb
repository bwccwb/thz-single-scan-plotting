# THz scan plotting

Works in **Google Colab** (Drive is mounted automatically) and locally (needs Google Drive for Desktop, or `THZ_DATA_DIR` set).
Just edit the `scans` list and the title fields below, then run.

```python
# Setup: run every session. In Colab this always installs the LATEST version from GitHub,
# so code changes you push show up here without restarting anything.
# If the repo is ever private, add a GitHub token as a Colab secret named GITHUB_TOKEN.
import sys
try:
    import google.colab
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

if IN_COLAB:
    repo = "github.com/bwccwb/thz-single-scan-plotting.git"
    try:
        from google.colab import userdata
        repo = f"{userdata.get('GITHUB_TOKEN')}@{repo}"
    except Exception:
        pass  # no token needed while the repo is public
    %pip install -q --upgrade --force-reinstall --no-deps "git+https://{repo}"

# forget any previously imported copy so the fresh install is used
for name in [m for m in sys.modules if m == "thzplot" or m.startswith("thzplot.")]:
    del sys.modules[name]
import thzplot

%matplotlib inline
print("Data folder:", thzplot.get_data_dir())
```

```python
# Optional: see what's on the Drive
print(thzplot.list_dates()[-10:])          # most recent date folders
print(thzplot.list_scans("2026.11.02"))   # scan times on one date
```

```python
#@title Plot settings { display-mode: "form" }
what = "power"  #@param ["power", "voltage"]
smooth = 2  #@param {type:"number"}
title = "Benchtop Test w/ Copper Waveguide"  #@param {type:"string"}
subtitle = "Resolution 1000 MHz, Time Constant 100 ms, {smoothing}, ~12cm Distance"  #@param {type:"string"}
publication = False  #@param {type:"boolean"}
water_lines = False  #@param {type:"boolean"}
save_as = ""  #@param {type:"string"}

# (date, time, legend label, optional color)
scans = [
    ("2026.10.02", "15.52.35", "final bench config 10/02"),
    ("2026.10.29", "16.53.27", "mounted in open fridge 10/29"),
    ("2026.11.02", "13.34.03", "4 K, ungrounded, 11/2"),
    ("2026.11.04", "16.19.24", "4 K, grounded, 11/4"),
]

fig, ax = thzplot.plot_scans(
    scans, what=what, smooth=smooth, title=title, subtitle=subtitle,
    publication=publication, water=water_lines, save=save_as or None,
)
```

`smooth` is the Gaussian σ in points (0 = raw data); `{smoothing}` in the subtitle is filled in automatically ("2σ Smoothing" / "No Smoothing").
Other options for `plot_scans`: `xlim=(100, 1000)`, `ylim=(10, 300)`, `logy=False`, `figsize=(9, 7)`, `legend=False`.
To save onto Drive from Colab use e.g. `save_as = "/content/drive/MyDrive/Figures/fig1.pdf"`.
If Drive couldn't be mounted (setup printed "Falling back to the Google Drive API"), save to e.g. `/content/fig1.pdf` instead and download it from the Files panel.
