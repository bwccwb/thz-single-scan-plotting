# THz scan plotting

Works in **Google Colab** (Drive is mounted automatically) and locally (needs Google Drive for Desktop, or `THZ_DATA_DIR` set).
Just edit the `scans` list and the title fields below, then run.

```python
# Setup: run once per session. In Colab this installs the package from GitHub.
try:
    import thzplot
except ImportError:
    %pip install -q git+https://github.com/bwccwb/thz-single-scan-plotting.git
    import thzplot

%matplotlib inline
print("Data folder:", thzplot.get_data_dir())
```

```python
# Optional: see what's on the Drive
print(thzplot.list_dates()[-10:])          # most recent date folders
print(thzplot.list_scans("2026.11.02"))   # scan times on one date
```

```python
#@title Plot settings { display-mode: "form" }
what = "smoothed"  #@param ["smoothed", "power", "voltage"]
title = "Benchtop Test w/ Copper Waveguide"  #@param {type:"string"}
subtitle = "Resolution 1000 MHz, Time Constant 100 ms, 2σ Smoothing, ~12cm Distance"  #@param {type:"string"}
publication = False  #@param {type:"boolean"}
water_lines = False  #@param {type:"boolean"}
save_as = ""  #@param {type:"string"}

# (date, time, legend label, optional color)
scans = [
    ("2026.10.02", "15.52.35", "final bench config 10/02"),
    ("2026.10.29", "16.53.27", "mounted in open fridge 10/29"),
    ("2026.11.02", "13.34.03", "4 K, ungrounded, 11/2"),
    ("2026.11.04", "16.19.24", "4 K, grounded, 11/4"),
]

fig, ax = thzplot.plot_scans(
    scans, what=what, title=title, subtitle=subtitle,
    publication=publication, water=water_lines, save=save_as or None,
)
```

Other options for `plot_scans`: `smooth=3`, `xlim=(100, 1000)`, `ylim=(10, 300)`, `logy=False`, `figsize=(9, 7)`, `legend=False`.
To save onto Drive from Colab use e.g. `save_as = "/content/drive/MyDrive/Figures/fig1.pdf"`.
